```python
import numpy as np
import matplotlib.pyplot as plt
class Neurone():
  def __init__(self,nb_entre):
    self.poid=np.random.randn(nb_entre)*0.01
    self.biais=np.random.randn()
    print(f"appel du constructeur de la classe neuron")
  def __str__(self):
     return f"{self.poid}  {self.biais}"
  def activation_relu(self,x):
    return np.maximum(0,x)
  def activation_sig(self,x):
    return 1/(1+np.exp(-x))
  def activation_tanh(self,x):
    return np.tanh(x)
  def propagation_avant(self,entre,activation="relu"):
    if len(entre)!=len(self.poid):
      raise ValueError("le nombre d\'entre n\'est pas compatible avant les poids ")
    sortie=self.poid@entre +self.biais # bon ici on pourrait utilise np.dot(self.poid entre) ce qui exactement evivalent a @ de numpy 
    if activation=="relu":
      z=self.activation_relu(sortie)
    elif activation=="sigmoid":
      z=self.activation_sig(sortie)
    elif activation=="tanh":
      z=self.activation_tanh(sortie)
    else:
      raise ValueError("fonction activation non reconnue")
    return z
  def prediction(self,entre,activation="sigmoid"):
    predict_proba_=self.propagation_avant(entre,activation)
    return 1 if predict_proba_>= 0.5 else 0

Neurone=Neurone(nb_entre=4)
x_sample=np.array([1.5,0.3,-0.4,2.0])
probabilite=Neurone.propagation_avant(x_sample,activation="sigmoid")
print(f"Sortie de la propagation avant est :{probabilite:.4f}")
class_predict=Neurone.prediction(x_sample)
print(f"la probabilite de la classe predicte est :{class_predict:.3f}")

```
