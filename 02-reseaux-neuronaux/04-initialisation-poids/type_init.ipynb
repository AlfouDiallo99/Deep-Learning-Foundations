Les differentes methode d'initatilisation pour la convergence du reseaux 

```python
#import des biblioteques a utilisé
import torch 
import torch.nn as nn
import math as m
```

```python

#initialisation Xavier avec Torch
def Xavier_init(layer):
  if isinstance(layer,nn.Linear):
    input_size=layer.in_features
    out_put_size=layer.out_features
    #l'ecartype pour le uniforme(Sigmoid/tanh)
    std=m.sqrt(2/(input_size+out_put_size))
    layer.weight.data.uniform_(-std,std)
    #verification si la couche a un biais au depart alors mettre toutes les valeurs à 0
    if layer.bias is not None:
      layer.bias.data.zero_()
```

```python
#initialisation de He avec torch
def he_init(layer):
  if isinstance(layer,nn.Linear):
    input_size=layer.in_features
    #output_size=layer.out_features pas besoin
    #l'ecartype pour la loi normale avec He(Relu/ses varientes)
    std=m.sqrt(2/input_size)
    layer.weight.data.normal_(0,std)
    if layer.bias is not None:
      layer.bias.data.zero_()
```

```python
#l'initialisation de lecun,c'est pour les reseaux tres prefonds
def lecun(layer):
  if isinstance(layer,nn.Linear):
    input_size=layer.in_features
    std=m.sqrt(1/input_size)
    layer.weight.data.normal_(0,std)
    if layer.bias is not None:
      layer.bias.data.zero_()
```
