```python
import numpy as np
#Données 
x,y=2,20
#parametres initiaux
w=1
b=0
#taux d'apprentissage 
leaning_rate=0.1
nb_iteration=10# Initialisation
x = 2
y = 10
w = 1
b = 0
taux_apprentissage = 0.01
nb_iteration = 20
historique_loss = []
# Entraînement
for i in range(nb_iteration):
  # 1. Forward
 z=w*x+b
 # 2. Calcul de la loss
 loss=(z - y)**2
 historique_loss.append(loss)
 # 3. Backpropagation
 dL_dz = 2 * (z - y)
 dL_dw = dL_dz * x
 dL_db = dL_dz
 # 4. Mise à jour des paramètres
 w=w-taux_apprentissage * dL_dw
 b=b-taux_apprentissage * dL_db
 print(f"Iteration {i+1}: " f"z={z:.4f}, "f"loss={loss:.4f}, "f"w={w:.4f}, "f"b={b:.4f}")


```

```python
#visualisation de la fonction de perte 
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
k = range(1, len(historique_loss) + 1)
plt.plot(k, historique_loss, marker="o")
plt.xlabel("Itération")
plt.ylabel("Loss")
plt.title("Évolution de la fonction de perte")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
#implementation d'un fonction simple pour ullustre la descente de gradient
#la fonction avec la quel on travail
def fonction(x):
  return 2*x**2 +x
#derivee de la fonction f(x)=2X^2+X
def derivee(x):
  return 4*x+1
#le gradient de la fonction
def gradient_descent(point_dep,taux_apprentissage ,nb_iteration):
  x=point_dep
  #creation d'un liste pour garder la trace de la progression du gradient,avec comme valeur initiale le point de depart 
  historique=[x]
  for i in range(nb_iteration):
    gradient=derivee(x)
    x=x-(gradient*taux_apprentissage)
    historique.append(x)
  #là on retourne le dernier pour et l'historique du gradient
  return x,historique
point_optimale,historique=gradient_descent(point_dep=5,taux_apprentissage=0.01,nb_iteration=50)
x_plot=np.linspace(-6,6,100)
plt.figure(figsize=(10,6))
plt.plot(x_plot,fonction(x_plot),'b-',label=f"f(x)=2X^2+X")
plt.plot(historique,[fonction(x) for x in historique],'ro-',label=f" progression ")
plt.xlabel("X")
plt.ylabel("f(x)")
plt.title("descente de gradient")
plt.show()

```

```python
x = 2
y = 10
w = 1
b = 0

taux_apprentissage = 0.01

# Première itération uniquement
z = w * x + b
loss = (z - y) ** 2

print("Avant mise à jour")
print("z =", z)
print("loss =", loss)

dL_dz = 2 * (z - y)
dL_dw = dL_dz * x
dL_db = dL_dz

w = w - taux_apprentissage * dL_dw
b = b - taux_apprentissage * dL_db

print("\nAprès mise à jour")
print("w =", w)
print("b =", b)

# Deuxième calcul
z = w * x + b
loss = (z - y) ** 2

print("\nDeuxième calcul")
print("z =", z)
print("loss =", loss)
```
