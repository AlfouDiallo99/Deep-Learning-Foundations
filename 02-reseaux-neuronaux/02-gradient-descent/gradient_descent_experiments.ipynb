```python
import numpy as np
import matplotlib.pyplot as plt
#implementation de la descente de gradient avec d'autre fonction
def fonction1(x):
  return x**5+x**4+x**3 -1
def derivee1(x):
    return 5*x**4 + 4*x**3 + 3*x**2
def gradient_descent1(point_dep,taux,nb_iteration):
  x=point_dep
  historique=[x]
  for i in range(nb_iteration):
    gradient=derivee1(x)
    x=x-(gradient*taux)
    historique.append(x)
  return x,historique


configuration=[{"depart":2.0,"taux":0.1,"iteration":100},
               {"depart":-1.0,"taux":0.05,"iteration":100},
               {"depart":0.0,"taux":0.01,"iteration":100}]
plt.figure(figsize=(10,6))
x_plot=np.linspace(-6,6,100)
plt.subplot(2,1,1)
plt.plot(x_plot,fonction1(x_plot),'b-',label=f"f(x)=x^5+X^4+X^3-1")
plt.grid(True)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("fonction et minimas locaux")
plt.subplot(2,1,2)
for i ,config in enumerate(configuration):
  point_optimale,historique=gradient_descent1(point_dep=config["depart"],taux=config["taux"],nb_iteration=config["iteration"])
  print(f"\nconfiguration{i+1}")
  print(f"point de depart:{config['depart']}")
  print(f"Taux d\'appentissage:{config['taux']}")
  print(f"nombre d\'itération:{config['iteration']}")
  print(f"Resultats finale :x={point_optimale:.4f} , f(x)={fonction1(point_optimale):.4f}")
  plt.plot(historique,label=f"Depart:{config['depart']}, taux:{config['taux']}")
plt.grid(True)
plt.xlabel("iteration")
plt.ylabel("x")
plt.title('convergence selon differentes configuration')
plt.tight_layout()
plt.show()
```
